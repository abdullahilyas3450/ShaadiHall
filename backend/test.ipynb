```python
model = ChatOpenRouter(
    model="google/gemini-3-flash-preview",
    temperature=0,
    max_retries=2,
    # other params...
)
```

```python
class ConverstationState(BaseModel):
    messages: Annotated[list[SystemMessage | HumanMessage | AIMessage], operator.add]
    hall_type: Optional[str] = None
    budget: Optional[str] = None
    location: Optional[str] = None
    date: Optional[str] = None
    capacity: Optional[int] = None

class structured_output(BaseModel):
    hall_name:str =  Field(description="the name of the hall")
    hall_location: str= Field(description="the location of the hall")
    hall_capacity: int = Field(description="the capacity of the hall")
    hall_price: str = Field(description="the price of the hall")
    hall_date: str = Field(description="the date of the hall")

```
